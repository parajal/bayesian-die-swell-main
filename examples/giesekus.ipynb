```python
import sys
import os
from pathlib import Path

CURRENT_DIR = os.getcwd()
FAXEN_ROOT = os.path.abspath(os.path.join(CURRENT_DIR, '..'))
sys.path.append(FAXEN_ROOT)
from packages import ROMCurve4BayesianInference

model = ROMCurve4BayesianInference(
    swell_root=Path(FAXEN_ROOT),
    model="giesekus", 
    train_data_rels=["datas/giesekus/uavg-0.1/train/curve4_y.txt"],
    mode="full_curve",  
    rom_method="gpr",
    scaler="minmax",
    eps=1e-6,
    sigma_noise_percent=0.01,
    lambda_bounds = (1.0, 10.0),
    beta_bounds = (0.1, 0.98),
    alpha_bounds = (0.1, 0.45),
    thin=1,
    sigma_bias=None,
    use_pressure = False,
    augment_eta0=False,
    pressure_filename= "pressure_all.txt",
    true_theta=[1.0, 0.1, 0.2])

model.load_data(filename="datas/oldroyd-rom3/wi-2/curve4_y.txt", 
                u_avg_obs=0.1, pressure_filename="datas/oldroyd-rom3/wi-2/pressure.txt")

model.plot_data()  
model.build_rom()
model.plot_prior()
model.run_mcmc(nwalkers=10, nsteps=10000, burn_fraction=0.5)
model.compute_N1()
model.plot_posterior_predictive()
model.plot_corner(true_param = False)
model.plot_trace_all()
```
